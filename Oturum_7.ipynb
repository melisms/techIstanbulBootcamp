```python
import csv
from pypdf import PdfReader
from langchain_core.documents import Document
```

```python
def pdf_yukle(yol, **ek_bilgi):
  return [Document(page_content=sayfa.extract_text(),                    # sayfanın metni,
                     metadata={"kaynak": yol, "sayfa": no, **ek_bilgi})     # kaynak, sayfa, ek bilgi.
            for no, sayfa in enumerate(PdfReader(yol).pages, start=1)]
```

```python
def metin_yukle(yol, **ek_bilgi):
  with open(yol, encoding="utf-8") as f:                                  # Dosyayı UTF-8 ile aç.
        return [Document(page_content=f.read(),                             # dosyanın tamamı,
                         metadata={"kaynak": yol, **ek_bilgi})]
```

```python
def csv_yukle(yol, **ek_bilgi):
  with open(yol, encoding="utf-8") as f:                                  # Dosyayı aç.
        return [Document(page_content="\n".join(f"{k}: {v}" for k, v in satir.items()),   # satır → metin,
                         metadata={"kaynak": yol, "satir": no, **ek_bilgi})              # hangi satır.
                for no, satir in enumerate(csv.DictReader(f), start=1)]
```

```python
import re
from langchain_text_splitters import (
    CharacterTextSplitter,
    RecursiveCharacterTextSplitter,
    MarkdownHeaderTextSplitter
)
```

```python
sabit = CharacterTextSplitter(separator="",chunk_size=300,chunk_overlap=50).split_documents(el_2026)
```

```python
ozyinelemeli = RecursiveCharacterTextSplitter(chunk_size=400,chunk_overlap=60).split_documents(el_2026)
```

```python
def bolumlere_ayir(dokumanlar):
  parcalar = []
  for d in dokumanlar:
    for bolum in re.split(r"\n(?=\d+\. [A-ZÇĞİÖŞÜ])", d.page_content):
      bolum = bolum.strip()
      if not re.match(r"\d+\. ", bolum):
        continue
      parcalar.append(Document(page_content=bolum,
                               metadata={**d.metadata,"bolum":bolum.split("\n")[0]}))
  return parcalar
```

```python
bolumler_2026 = bolumlere_ayir(el_2026)
```

```python
sss_parcalari = MarkdownHeaderTextSplitter(
    headers_to_split_on=[("#", "baslik"), ("##", "bolum")],        # Başlık düzeyi → metadata alanı.
).split_text(bt_sss[0].page_content)
```

```python
for p in sss_parcalari:                                             # Her SSS parçasına
    p.metadata.update(bt_sss[0].metadata)
```

```python
import numpy as np
```

```python
def ilk_rag(soru,parcalar,k=3):
  metinler = [p.page_content for p in parcalar]
  V = mpnet.encode(metinler, normalize_embeddings=True)
  q = mpnet.encode(soru, normalize_embeddings=True)
  benzerlik = V @ q
  en_iyi = np.argsort(-benzerlik)[:k]
  for i in en_iyi:
    print(f"{benzerlik[i]:.2f} | {parcalar[i].metadata['bolum']}")
  baglam = "\n\n".join(metinler[i] for i in en_iyi)
  return sor(f"Bağlam:\n{baglam}\n\nSoru:{soru}",
             sistem="Yalnızca bağlama göre cevapla; bağlamda yoksa 'Dokümanda yok' de."
             )
```

#RAG Pipeline

```python
from langchain_core.runnables import RunnableLambda, RunnablePassthrough
```

```python
def bicimle(docs):
  return "\n\n".join(
      f"[{d.metadata['kaynak']}, {d.metadata['bolum']}] {d.page_content}"
      for d in docs
  )
```

```python
KURALLAR = ("Yalnızca verilen bağlama göre cevapla."
            "Bağlamda yoksa 'Bu bilgi dokümanda yok.' yaz."
            "Sonuna kaynağı (dosya, bölüm) ekle."
)
```

```python
def cevapla(girdi):
  return sor(f"Bağlam:\n{girdi['baglam']}\n\nSoru: {girdi['soru']}",sistem=KURALLAR)
```

```python
rag_zinciri = (
    {"baglam":filtreli | RunnableLambda(bicimle),"soru": RunnablePassthrough()} | RunnableLambda(cevapla)
)
```

```python
for soru in ["Tatile kaç gün çıkabilirim?","Evden şirket ağına nasıl bağlanırım?","Şirketin genel müdür kim?"]:
  print("Soru: ",soru)
  print("Cevap: ",rag_zinciri.invoke(soru).strip())
```

```python
import datetime
import requests
import pandas as pd
from langchain_core.tools import tool
```

```python
calisanlar = pd.read_csv("calisanlar.csv")
KOORDINAT = {"istanbul":(41.01,28.98), "ankara":(39.93,32.85),
             "izmir":(38.42,27.14)
}
```

```python
arayici = vs.as_retriever(search_kwargs={"k":3,"filter":{"yil":2026}})
```

```python
@tool
def el_kitabinda_ara(soru:str) -> str:
  """El kitabında bir soruya cevap arar."""
  return bicimle(arayici.invoke(soru))

@tool
def calisan_bilgisi(ad_soyad:str) -> str:
  """Bir çalışanın adını ve soyadını alarak o çalışanın bilgilerini döndürür."""
  satir = calisanlar[calisanlar["ad_soyad"]==ad_soyad.strip()]
  if satir.empty:
    return f"'{ad_soyad}' adlı çalışan bulunamadı."
  return satir.iloc[0].to_json(force_ascii=False)

@tool
def hesapla(ifade:str) -> str:
  """Verilen matematiksel ifadeyi değerlendirir ve sonucunu döndürür."""
  if not re.fullmatch(r"[\d\s+\-*().]+",ifade):
    return "Geçersiz ifade."
  return str(eval(ifade))

@tool
def hava_durumu(sehir: str) -> str:
  """Verilen şehir için mevcut hava durumu bilgisini (sıcaklık) döndürür."""
  konum = KOORDINAT.get(sehir.replace("İ","i").lower())
  if not konum:
    return f"'{sehir}' için konum bilgim yok."
  try:
    hava = requests.get("https://api.open-meteo.com/v1/forecast",
                        params={"latitude":konum[0],"longitude":konum[1],
                                "current":"temperature_2m"},timeout=10).json()
    return f"{sehir}: {hava['current']['temperature_2m']} derecedir."
  except:
    return f"Hava servisine ulaşılamadı."
```

```python
ARACLAR = {a.name: a for a in [el_kitabinda_ara, calisan_bilgisi, hesapla, hava_durumu]}
```

```python
AJAN_TALIMATI = f"""Sen Bulutkedi Yazılım A.Ş.'nin İK asistanısın; soruları araçlarla çözersin.
Bugünün tarihi: {datetime.date.today():%d-%m-%Y (%A)}.
Araçlar:
{arac_listesi}
Her adımda YALNIZCA tek bir JSON nesnesi yaz:
- Araç çağırmak için: {{"dusunce":"neden","arac":"arac_adı","argumanlar":{{"parametre":"deger"}}}}
- Son cevap için: {{"dusunce":"neden","cevap":"..."}}
Kurallar: Şirket kurallarını, çalışan bilgilerini ve havayı bilmezsin; araçla öğren, uydurma.
Her hesabı hesapla aracıyla yap. Tek seferde tek araç çağır. Kimlik gerekiyorsa ve bilinmiyorsa sor.
"""
```

```python
class Karar(BaseModel):
  dusunce: str = ""
  arac: str | None = None
  argumanlar: dict = {}
  cevap: str | None = None
```

```python
hafiza = {}
```

```python
def ajan(soru, oturum="varsayilan",maks_adim=10):
  gecmis = hafiza.setdefault(oturum, [])
  gecmis.append(f"Kullanıcı: {soru}")
  for _ in range(maks_adim):
    try:
      karar = json_coz(sor("\n".join(gecmis),sistem=AJAN_TALIMATI),Karar)
    except Exception as hata:
      gecmis.append(f"Sistem: cevabın geçersizdi ({type(hata).__name__});"
      "yalnızca JSON yaz.")
      continue
    gecmis.append(f"Asistan: {karar.model_dump_json(exclude_none=True)}")
    if karar.cevap or karar.arac not in ARACLAR:
      print("Cevap:",karar.cevap)
      return

    try:
      sonuc = ARACLAR[karar.arac].invoke(karar.argumanlar)
    except Exception as hata:
      sonuc = f"Araç hatası: {hata}"
    print(f" -> {karar.arac}({karar.argumanlar}) = {str(sonuc)}")
    gecmis.append(f"Araç sonucu {karar.arac}: {sonuc}")
  print("Adım sınırına ulaşıldı.")
```
